```python
import pandas as pd

from kbr import paths, profile

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 200)
```

## 1. Wczytanie

Ścieżki wyłącznie przez `paths.raw()`. Kodowanie i separator sprawdź na kilku pierwszych bajtach,
zanim wczytasz całość — pliki od wykonawców bywają w cp1250.

```python
sciezka = paths.raw("kbr2018", "...", "...")
print(sciezka, sciezka.exists())
print("sha256:", paths.checksum(sciezka))

df = pd.read_csv(sciezka, sep=";", encoding="cp1250")
df.shape
```
